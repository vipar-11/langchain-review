# Langchain Expression Language (LCEL)

```python
import os
from dotenv import load_dotenv
```

## Load API Keys

```python
load_dotenv()

LLM_MODEL = os.getenv('LLM_MODEL')
LLM_BASE_URL = os.getenv('LLM_BASE_URL')
GROQ_API_KEY = os.getenv('GROQ_API_KEY')
```

```python
LLM_MODEL, LLM_BASE_URL, GROQ_API_KEY[:5]
```

## Initialize the LLM Client

```python
import langchain
from langchain.chat_models import init_chat_model
```

```python
model_provider = LLM_MODEL.split('/')[0]
model_provider
```

```python
llm = init_chat_model(
    model=LLM_MODEL,
    model_provider=model_provider,
    api_key=GROQ_API_KEY,
    base_url=LLM_BASE_URL
)
```

## Make an LLM call using LCEL

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt = ChatPromptTemplate.from_template("What is the capital of {country}?")
```

```python
chain = prompt | llm
```

```python
response = chain.invoke({'country': 'US'})
print(response.content)
```

```python
response = chain.invoke({'country': 'France'})
print(response.content)
```

## Works for structured outputs as well

```python
from pydantic import BaseModel, Field
```

```python
class Capital(BaseModel):
    country: str = Field(description='Name of the country')
    capital: str = Field(description='Name of capital city')
```

```python
llm_with_structured_output = llm.with_structured_output(Capital)
```

```python
chain = prompt | llm_with_structured_output
```

```python
response = chain.invoke({'country': 'India'})
print(response.model_dump_json(indent=2))
```
